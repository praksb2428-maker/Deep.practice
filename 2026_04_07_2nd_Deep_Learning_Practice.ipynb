```python
# 데이터 사전
def create_fixed_dictionary(df, columns):
    data_dict = pd.DataFrame({

        '컬럼명': columns,
        '데이터타입': [df[col].dtype for col in columns],
        '고유값수': [df[col].nunique() for col in columns],
        '샘플값': [str(df[col].iloc[0]) for col in columns],
        '설명': [
            '📅 관측일자 (YYYY-MM-DD)',
            '🌡️ 평균기온 (°C)',
            '💧 상대습도 (%)',
            '💨 풍속 (km/h)',
            '⛅ 평균기압 (hPa)'
        ][:len(columns)]
    })
    return data_dict

display(data_dict)
```
