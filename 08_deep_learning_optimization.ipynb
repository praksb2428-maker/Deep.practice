# 13장 모델의 성능 검증하기

##  1. 데이터의 확인과 예측 실행

```python
import pandas as pd

# 광물 데이터 로드
!git clone https://github.com/taehojo/data.git
df = pd.read_csv('./data/sonar3.csv', header=None)
df.head()
```

```python
#  60번째 열에 이진분류로 나뉘어진 데이터의 갯수를 출력
df[60].value_counts()
```

```python
X = df.iloc[:,0:60] # 모든 행의 0~59행 열을 선택
y = df.iloc[:,60] # 모든 행의 60번 열 선택
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 모델 설정
model = Sequential() # 층을 순서대로 쌓는 모델 생성
model.add(Dense(24,  input_dim=60, activation='relu')) # 순서. 뉴런의 갯수, 입력층, 첫번째 은닉층
model.add(Dense(10, activation='relu')) # 두번쨰 은닉층
model.add(Dense(1, activation='sigmoid')) # 출력층
# 손실함수, 최적화 알고리즘, 평가지표를 사용하겠다는 코드
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# x와y를 학습을 200번 반복하고 10개씩 끊어서 문제를 확인
history=model.fit(X, y, epochs=200, batch_size=10)
```

## 3. 학습셋과 테스트셋

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split

import pandas as pd
```

```python
# 학습셋과 테스트셋 구분
# test에 0.3을 정의하고 나머지 0.7이 train으로 나뉘어졌음
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, shuffle=True)
```

```python
# 모델 설정
model = Sequential()
model.add(Dense(24,  input_dim=60, activation='relu'))
model.add(Dense(10, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# 나뉘어진 70%의 train으 데이터만으로 학습을 200번 반복하고 10개씩 끊어서 문제를 확인
history=model.fit(X_train, y_train, epochs=200, batch_size=10)
```

```python
# test의 데이터를 이용하여 성능평가
score=model.evaluate(X_test, y_test)
print('Test accuracy:', score[1])
```

## 5. k겹 교차 검증

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score

import pandas as pd
```

```python
# fold 설정
k=5 # 5개의 테스트를 진행
kfold = KFold(n_splits=k, shuffle=True) # 데이터를 5번을 나눔

# 5번의 테스트의 결과를 저장
acc_score = []

#한번의 테스트를 진행하고 다음 테스트를 진행하기위해 초기화하고 다시 실행하는 코드
def model_fn():
    model = Sequential()
    model.add(Dense(24, input_dim=60, activation='relu'))
    model.add(Dense(10, activation='relu'))
    model.add(Dense(1, activation='sigmoid'))
    return model

# K겹 교차 검증을 이용해 k번 학습 실행
for train_index , test_index in kfold.split(X):  # for 문 k번 반복. spilt()에 의해 k개의 학습셋, 테스트셋으로 분리
    X_train , X_test = X.iloc[train_index,:], X.iloc[test_index,:]
    y_train , y_test = y.iloc[train_index], y.iloc[test_index]

    model = model_fn()
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    history=model.fit(X_train, y_train, epochs=200, batch_size=10, verbose=0)

    accuracy = model.evaluate(X_test, y_test)[1] # 정확도
    acc_score.append(accuracy)

# k번 실시된 정확도 평균
avg_acc_score = sum(acc_score)/k

# 결과 출력
print('정확도:', acc_score)
print('정확도 평균:', avg_acc_score)
```
