```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
raw = load_breast_cancer()
df = pd.DataFrame(data = raw.data, columns = raw.feature_names)
```

```python
df.head()
```

```python
df.value_counts()
```

```python
colormap = plt.cm.gist_heat   # 그래프 색상 구성
plt.figure(figsize=(500,500))
```

```python
sns.heatmap(df.corr(),linewidths=0.1,vmax=0.5, cmap=colormap, linecolor='white', annot=True)
plt.show()
```

worst symmetry와 mean radius의 연관관계가 하얀색으로 표기되어 연관관계가 높은걸 알 수 있다.

worst perimeter와 smoothness error의 연관관계가 검정색으로 표기되어 연간관계가 낮은걸 알 수 있다.

하얗게 나올수록 연관관계가 높고 검정으로 나올수록 연관관계가 낮게 나오는것 을확인할 수 있었다.
