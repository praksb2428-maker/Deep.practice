```python
import pandas as pd
import numpy as np
from google.colab import drive
from google.colab import files
import os

# 1. 구글 드라이브 마운트
drive.mount('/content/drive')

# [수정 필요] 파일이 저장된 구글 드라이브 경로를 입력하세요.
# 예: '/content/drive/MyDrive/여름철_기후데이터_날짜정리.xlsx'
file_path = '/content/drive/MyDrive/2010-2025 Summer Temperature Data.xlsx'

if not os.path.exists(file_path):
    print(f"파일을 찾을 수 없습니다: {file_path}")
else:
    # 2. 데이터 로드
    df = pd.read_excel(file_path)

    # 3. NumPy를 활용한 강수량 NULL 처리 및 데이터 추출
    # 강수량 컬럼의 NaN(NULL) 값을 NumPy를 사용하여 0으로 변환합니다.
    prec_raw = df['Precipitation(mm)'].values
    prec_array = np.nan_to_num(prec_raw, nan=0.0)

    avg_array = df['Average Temperature(℃)'].values
    max_array = df['Maximum(℃)'].values

    # 4. 평균온도 라벨링 (NumPy 벡터 연산)
    # 30도 이상: 2 / 25~30도: 1 / 25도 미만: 0
    avg_labels = np.zeros_like(avg_array, dtype=int)
    avg_labels[avg_array >= 30] = 2
    avg_labels[(avg_array >= 25) & (avg_array < 30)] = 1
    # 25도 미만은 초기값 0 유지

    # 5. 최고온도 라벨링 (NumPy 벡터 연산)
    # 35도 이상: 2 / 29~35도: 1 / 29도 미만: 0
    max_labels = np.zeros_like(max_array, dtype=int)
    max_labels[max_array >= 35] = 2
    max_labels[(max_array >= 29) & (max_array < 35)] = 1
    # 29도 미만은 초기값 0 유지

    # 6. 강수량 라벨링 (NumPy 벡터 연산)
    # 70mm 이상: 2 / 50~70mm: 1 / 50mm 미만: 0
    prec_labels = np.zeros_like(prec_array, dtype=int)
    prec_labels[prec_array >= 40] = 2
    prec_labels[(prec_array >= 20) & (prec_array < 40)] = 1
    # 50mm 미만은 초기값 0 유지

    # 7. 결과 통합 및 데이터프레임 업데이트
    df['Precipitation(mm)'] = prec_array  # 결측치가 0으로 채워진 데이터
    df['Avg_Temp_Level'] = avg_labels
    df['Max_Temp_Level'] = max_labels
    df['Precip_Level'] = prec_labels

    # 8. CSV 파일 저장 및 다운로드
    output_filename = 'processed_summer_weather_data.csv'
    df.to_csv(output_filename, index=False, encoding='utf-8-sig')

    print("전처리 및 라벨링이 완료되었습니다.")
    print(f"생성된 파일: {output_filename}")

    # 코랩 브라우저로 파일 자동 다운로드
    files.download(output_filename)
```
